# Syntax detective
**Module 1, Lesson 1: How Python runs your code.** **Solution notebook.** Reference solution. Try the starter notebook first.

```python
import keyword
SNIPPETS = {
 'good': 'x = 5\nif x > 3:\n    print(x)',
 'missing colon': 'if x > 3\n    print(x)',
 'bad indent': 'x = 1\n  print(x)',
 'keyword as name': 'class = 10',
 'case matters': 'Total = 1\ntotal = 2\nprint(Total, total)',
}
print(len(keyword.kwlist), 'keywords, for example', keyword.kwlist[:5])
```

## Step 1. Valid or not?

```python
def check(code):
    # TODO: return 'OK' if compile(code, '<lab>', 'exec') works, else the error class name (e.g. 'SyntaxError')
    try:
        compile(code, '<lab>', 'exec')
        return 'OK'
    except SyntaxError as e:
        return type(e).__name__

for name, code in SNIPPETS.items():
    print(f'{name:16} {check(code)}')
```

## Step 2. Valid variable names

```python
def is_valid_name(name):
    # TODO: True if name.isidentifier() and it is not a keyword
    return name.isidentifier() and not keyword.iskeyword(name)

for n in ['total', '2nd_place', 'first_name', 'class', '_hidden', 'my-name', 'Total']:
    print(f'{n:12} {is_valid_name(n)}')
```
