## Step 1. Whole-dollar split

```python
def split(total, people):
    # TODO: return (each, left_over) using // and %
    return total // people, total % people

each, left = split(47, 6)
print(f'each {each}', f'left over {left}', sep=' | ', end='\n')
```

## Step 2. Split to the cent
Work in cents. The first `left` people pay one extra cent.

```python
def cents_split(total_dollars, people):
    # TODO: convert to cents with round(total_dollars * 100); each = cents // people; left = cents % people; return a list of shares in dollars, first `left` shares get 1 extra cent
    cents = round(total_dollars * 100)
    each, left = cents // people, cents % people
    return [(each + (1 if i < left else 0)) / 100 for i in range(people)]

shares = cents_split(100.00, 3)
print(*shares, sep=' + ', end=' = ')
print(round(sum(shares), 2))
```
