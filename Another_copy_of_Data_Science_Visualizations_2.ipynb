# **Importing the PANDAS & MATPLOTLIB packages + Read in the Data**



```python
#import the pandas package
import pandas as pd

#import the matplotlib package
import matplotlib.pyplot as plt

#read in the data
import gdown
url = 'https://drive.google.com/file/d/13KYOUR5-FjuDDaa_NJD2LyPR7EKncOfn/view?usp=sharing'
output_path = 'netflix_data.csv'
gdown.download(url, output_path, quiet=False, fuzzy=True)
netflix = pd.read_csv('netflix_data.csv')
print(netflix)
print(len(netflix))#displays how many rows are in the dataset

netflix = netflix.dropna()#removes rows with empty data (a version of cleaning) print(len(netflix))#displays how many rows are in the dataset after cleaning

#HOW MANY ROWS WERE DROPPED?
```

# **Basic Histogram**
●	Histogram - A histogram is a graphical representation that shows the frequency of numerical data organized into user-speciﬁed ranges (bins or buckets).

[Colors](https://matplotlib.org/stable/gallery/color/named_colors.html)



```python
#pull out the years from the netflix content
years = netflix['Year'] #sub dataset print(years)

#create a histogram with the years
plt.hist(years)

#label the graph
plt.title('Years Distribution of Netflix Content')
plt.xlabel('Year')
plt.ylabel('Count')
plt.show()
```

# **Histogram [Documentation](https://matplotlib.org/stable/api/_as_gen/matplotlib.axes.Axes.hist.html#matplotlib.axes.Axes.hist)**

Use your favorite color and change the code of the graph




```python
#change the amount of bins, color, and edgecolor
plt.hist(years, bins = 20, color = 'purple', edgecolor = 'black')
#CHANGE THE COLOR TO YOUR FAVORITE COLOR
```

# **Basic Scatter Plot**

**Scatter Plot** - A scatter plot shows the relationship between two numerical attributes.



```python
#let's compare runtimes to IMDb Ratings for the 2020 movies

movies2020 = netflix[netflix['Year']==2020] #sub dataset of 2020 movies
print(movies2020)

runtimes2020 = movies2020['Runtime'] #runtime values from 2020 movies sub dataset
print(runtimes2020)

ratings2020 = movies2020['IMDb'] #IMDb ratings from 2020 movies sub dataset
print(ratings2020)

#create scatter plot
plt.scatter(runtimes2020,ratings2020)
plt.title('Runtimes VS IMDb Ratings of 2020 Netflix Content')
plt.xlabel('Runtime (in minutes)')
plt.ylabel('IMDb Rating')
plt.show()

```


# **Basic Pie Chart**
●	 **Pie Chart** - A pie chart (or a pie graph) is a circular graphical chart, which is divided into slices in order to illustrate numerical proportions of categorical data.


```python
#Let's make a pie chart to show the breakdown of age groups of the netflix content
#need to count the amount of each age group using the line below
age_counts = netflix['Age'].value_counts()

#create a pie chart using the age counts and their labels, rounding 2 decimal places
plt.pie(age_counts, labels = age_counts.index, autopct="%.2f%%")
plt.title("Ages of Netflix Content")
plt.legend()
plt.show()
```

# **Pie Chart [Documentation](https://matplotlib.org/stable/api/_as_gen/matplotlib.axes.Axes.pie.html)**
